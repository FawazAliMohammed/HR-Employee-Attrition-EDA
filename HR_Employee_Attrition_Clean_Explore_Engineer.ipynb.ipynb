```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

# Show the first 5 rows
df.head()
```

```python
# Check number of rows and columns
df.shape
```

```python
# See all column names
df.columns
```

```python
# Check the data types
df.dtypes
```

```python
# Get basic information about the dataset
df.info()
```

```python
# Check how many missing values are in each column
df.isnull().sum()
# Check total number of missing values
df.isnull().sum().sum()
```

```python
# Check how many duplicate rows we have
df.duplicated().sum()
```

```python
# Remove duplicate rows if there are any
df = df.drop_duplicates()

# Check the shape again
df.shape
```

```python
# Get statistics for numerical columns
df.describe()
```

```python
# Count how many employees stayed and how many left
sns.countplot(x="Attrition", data=df)

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Find the percentage of employees who left
attrition_percentage = df["Attrition"].value_counts(normalize=True) * 100

attrition_percentage
```

```python
# Plot the distribution of employee age
plt.hist(df["Age"], bins=10)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Compare monthly income for employees who stayed and left
sns.boxplot(x="Attrition", y="MonthlyIncome", data=df)

plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.show()
```

```python
# Select only numerical columns
numeric_data = df.select_dtypes(include=["int64", "float64"])

# Calculate correlation
correlation = numeric_data.corr()

# Create heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(correlation, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Count employees based on overtime and attrition
pd.crosstab(df["OverTime"], df["Attrition"])

# Create a bar chart
sns.countplot(x="OverTime", hue="Attrition", data=df)

plt.title("Attrition Based on Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Compare job satisfaction with attrition
pd.crosstab(df["JobSatisfaction"], df["Attrition"])

# Create a graph
sns.countplot(x="JobSatisfaction", hue="Attrition", data=df)

plt.title("Job Satisfaction and Attrition")
plt.xlabel("Job Satisfaction")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Create a new feature
# It shows how much of an employee's total working experience
# has been spent at the current company.

df["CompanyExperienceRatio"] = (
    df["YearsAtCompany"] / df["TotalWorkingYears"]
)

# Replace any infinity or missing values with 0
df["CompanyExperienceRatio"] = df["CompanyExperienceRatio"].replace(
    [np.inf, -np.inf], 0
)

df["CompanyExperienceRatio"] = df["CompanyExperienceRatio"].fillna(0)

# Show the new column
df[["YearsAtCompany", "TotalWorkingYears", "CompanyExperienceRatio"]].head()
```

```python
# Create another new feature
# It compares years since last promotion with total working years.

df["PromotionRatio"] = (
    df["YearsSinceLastPromotion"] / df["TotalWorkingYears"]
)

# Replace missing or infinite values
df["PromotionRatio"] = df["PromotionRatio"].replace(
    [np.inf, -np.inf], 0
)

df["PromotionRatio"] = df["PromotionRatio"].fillna(0)

# Show the new feature
df[["YearsSinceLastPromotion", "TotalWorkingYears", "PromotionRatio"]].head()
```

```python
# Create income per job level
df["IncomePerJobLevel"] = df["MonthlyIncome"] / df["JobLevel"]

# Show the result
df[["MonthlyIncome", "JobLevel", "IncomePerJobLevel"]].head()
```

```python
# Display the new features
df[[
    "CompanyExperienceRatio",
    "PromotionRatio",
    "IncomePerJobLevel"
]].head(10)

# Check the final number of rows and columns
df.shape
```

```python
## 5 Key Insights

#1.The dataset contains more employees who stayed with the company than employees who left, so the Attrition variable is imbalanced.
#2.Employees who work overtime appear to have a higher chance of leaving the company. This suggests that overtime may be an important factor related to employee attrition.
#3.The age distribution shows that most employees are in the middle-age range, with fewer employees at the very young and very old ends.
#4.Monthly income is different between employees who stayed and employees who left. This suggests that compensation may have some relationship with attrition.
#5.Job satisfaction also shows differences between employees who stayed and employees who left. Therefore, job satisfaction could be useful as a feature in a future employee attrition prediction model.
```
